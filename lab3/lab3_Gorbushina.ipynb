Выполните следующие задания:

1. Загрузите датафрейм из файла polit.csv и сохраните его в переменную polit. Если в датафрейме есть строки с пропущенными значениями, то удалите их.
2. Выберите в датафрейме строки, которые соответствуют странам со значениями индекса Freedom House (fh09) выше 5.
3. Выберите в датафрейме строки, которые соответствуют странам Африки (afri) с процентом женщин в парламенте (fparl08) выше 30%.
4. Выберите в датафрейме строки, которые соответствуют странам Африки или Латинской Америки (afri и lati) со значением индекса Polity2 (polity09) больше или равным 8.
5. Добавьте в датафрейм polit столбец corr_round, в котором будут храниться округленные до 2 знака после запятой значения индекса Control of Corruption (corr0509).
6. Добавьте в датафрейм polit столбец fh_status, в котором будут храниться типы стран в зависимости от значения индекса Freedom House (значения free, partly free, not free). Соответствие значений fh09 с типами стран см. здесь https://freedomhouse.org/sites/default/files/2020-02/Methodology_FIW_2016.pdf (Table 3 на последней странице).
7. Сгруппируйте строки в таблице в соответствии со значениями столбца fh_status, полученного в предыдущей части и выведите минимальное, среднее и максимальное значение показателя gini (индекс Джини) по каждой группе.
8. Сгруппируйте строки в таблице в соответствии со значениями столбца fh_status и запишите строки, относящиеся к разным группам, в отдельные csv-файлы.

````python
import pandas as pd
import numpy as np

# 1 (грузим, чистим)
polit = pd.read_csv('polit.csv', decimal=',')
key_columns = ['fh09', 'polity09', 'fparl08', 'corr0509']
polit = polit.dropna(subset=key_columns)

def format_countries_list(countries, countries_per_line=5):
    if len(countries) == 0:
        return "-"
    
    result = []
    for i in range(0, len(countries), countries_per_line):
        chunk = countries[i:i + countries_per_line]
        result.append(', '.join(chunk))
    return '\n'.join(result)

# 2
fh_above_5 = polit[polit['fh09'] > 5]
print("2. Страны с индексом Freedom House (fh09) > 5:")
print(format_countries_list(fh_above_5['ctry'].tolist()))
print()

# 3

africa_women = polit[(polit['afri'] == 1) & (polit['fparl08'] > 30)]
print("3. Африканские страны с долей женщин в парламенте > 30%:")
print(format_countries_list(africa_women['ctry'].tolist()))
print()

# 4

africa_latin_democracy = polit[((polit['afri'] == 1) | (polit['lati'] == 1)) & (polit['polity09'] >= 8)]
print("4. Страны Африки или Латинской Америки с Polity2 >= 8:")
print(format_countries_list(africa_latin_democracy['ctry'].tolist()))
print()

# 5
polit['corr_round'] = polit['corr0509'].round(2)
print("5. + столбец corr_round (округл. знач. corr0509 до 2 знаков):")
print(polit[['ctry', 'corr0509', 'corr_round']].head().to_string(index=False))
print()

# 6
def get_fh_status(fh_value):
    if fh_value <= 2.5:
        return 'free'
    elif fh_value <= 5.0:
        return 'partly free'
    else:
        return 'not free'
polit['fh_status'] = polit['fh09'].apply(get_fh_status)
print("6. Добавлен столбец fh_status (типы стран по Freedom House)")
print("Первые 10 строк для примера:")
print(polit[['ctry', 'fh09', 'fh_status']].head(10).to_string(index=False))
print("\n")

# 7
gini_stats = polit.groupby('fh_status')['gini'].agg(['min', 'mean', 'max']).round(2)
print("7. Статистика индекса Джини по группам стран:")
print(gini_stats)
print("\n")

# 8
fh_status_groups = polit.groupby('fh_status')
print("8. Сохранение групп в CSV-файлы:")
for status, group in fh_status_groups:
    filename = f"polit_{status.replace(' ', '_')}.csv"
    group.to_csv(filename, index=False)
    print(f"Файл: {filename}")
print("Все группы сохранены в отдельные CSV-файлы")
````

5. Вывести жанры игр с количеством игр, которые принадлежат соответствующему жанру в виде "<жанр> - <количество игр>" для всех жанров, которые содержат в своем названии не менее 3 различных гласных.

````python
import pandas as pd
import numpy as np

vg_data = pd.read_csv('vgsales.csv')
metacritic_data = pd.read_csv("metacritic_games.csv")

# 1
platforms = vg_data['Platform'].unique()
print(f"\n1. Все доступные платформы, на которых выпускалась хотя бы одна игра: {len(platforms)}")
for i, platform in enumerate(sorted(platforms), 1):
    print(f"{i:2d}. {platform}")
print()

# 2
def get_metacritic_rating(game_name):
    try:
        if not metacritic_data.empty and 'name' in metacritic_data.columns:
            match = metacritic_data[
                metacritic_data['name'].str.contains(str(game_name), case=False, na=False, regex=False)
            ]
            if not match.empty:
                return match.iloc[0]['metascore']
    except Exception as e:
        pass
    return np.nan

vg_data['metacritic_rating'] = vg_data['Name'].apply(get_metacritic_rating)
print("2. Добавлен столбец metacritic_rating, пример (10 строк):")
print(vg_data[['Name', 'Platform', 'Year', 'metacritic_rating']].head(10).to_string(index=False))
print()

# 3
print('3. Игры с рейтингом "M" и год издания не ранее 2012:')
def get_esrb_rating(game_name):
    try:
        if not metacritic_data.empty and 'name' in metacritic_data.columns and 'rating' in metacritic_data.columns:
            match = metacritic_data[
                metacritic_data['name'].str.contains(str(game_name), case=False, na=False, regex=False)
            ]
            if not match.empty:
                return match.iloc[0]['rating']
    except:
        pass
    return np.nan
vg_data['esrb_rating'] = vg_data['Name'].apply(get_esrb_rating)

m_rated_games = vg_data[
    (vg_data['esrb_rating'] == 'M') & 
    (vg_data['Year'] >= 2012) & 
    (vg_data['Year'].notna())
]

print(f"Количество игр с рейтингом 'M' (2012+): {len(m_rated_games)}")
if len(m_rated_games) > 0:
    print("Первые 10:")
    print(m_rated_games[['Name', 'Platform', 'Year', 'esrb_rating', 'Genre']].head(10).to_string(index=False))
    
# 4
print("4. Описательные статистики для игр с рейтингом 'M' (2012+):")
if len(m_rated_games) > 0:
    sales_columns = ['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']
    
    print("Статистики по продажам (в миллионах):")
    sales_stats = m_rated_games[sales_columns].describe().round(2)
    print(sales_stats)
    
    print("\nСтатистики по годам выпуска:")
    year_stats = m_rated_games['Year'].describe()
    print(year_stats)
    
    print("\nРаспределение по жанрам:")
    genre_dist = m_rated_games['Genre'].value_counts()
    print(genre_dist)
    
    print("\nРаспределение по платформам:")
    platform_dist = m_rated_games['Platform'].value_counts()
    print(platform_dist)
    
    print("\nСтатистики по рейтингу Metacritic:")
    if m_rated_games['metacritic_rating'].notna().any():
        metacritic_stats = m_rated_games['metacritic_rating'].describe()
        print(metacritic_stats)
    else:
        print("Нет данных по рейтингу Metacritic")
print()

# 5
print("5. Жанры игр с количеством игр (жанры с ≥3 различными гласными):")
def count_unique_vowels(text):
    if pd.isna(text):
        return 0
    vowels = set('aeiou') 
    text_lower = str(text).lower()
    unique_vowels = {char for char in text_lower if char in vowels}
    return len(unique_vowels)

genre_counts = vg_data['Genre'].value_counts()
filtered_genres = {}
for genre, count in genre_counts.items():
    vowel_count = count_unique_vowels(genre)
    if vowel_count >= 3:
        filtered_genres[genre] = count

if filtered_genres:
    print(f"Найдено жанров: {len(filtered_genres)}")
    for genre, count in filtered_genres.items():
        print(f"  {genre} - {count} игр")
````
