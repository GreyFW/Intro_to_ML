```
def sinplot(flip=1):
    x = np.linspace(0, 14, 100)
    for i in range(1, 5):
        plt.plot(x, np.sin(x + i * .5) * (7 - i) * flip)

sb.set_style("darkgrid", {'axes.axisbelow': True}) # Изменили здесь
sinplot()

```
Уточнение: `'axes.axisbelow'` может принимать два
значения: `False` и `True`.

`````python
def task1_9():
    def sinplot(flip=1):
        x = np.linspace(0, 14, 100)
        for i in range(1, 5):
            plt.plot(x, np.sin(x + i * .5) * (7 - i) * flip)

    sb.set_style("darkgrid", {'axes.axisbelow': True}) # Изменили здесь
    sinplot()
    sb.despine()
    plt.show()
    
task1_9()
`````

**Отличие в графиках** состоит в том, что при параметре `axisbelow: True` сетка координат рисуется **под** графиком, а при `False` над графиком, т.е. линии будут поверх изображения.
