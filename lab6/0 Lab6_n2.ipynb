### **РЕШЕНИЕ**

Примечание: буду копировать куски (пункты) из условия выше для удобства

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LassoCV
```

1. Загрузите данные в DataFrame, используя функцию read_csv.

```python
df = pd.read_csv('house_price.csv')
```

2. Сколько строк и столбцов в данных? Есть ли пропуски? Для ответа на
вопросы используйте метод info().

```python
df.shape
```

```python
df.info()
```

3. Для выполнения задания в наборе данных необходимо оставить только
числовые признаки. Для этого можно использовать следующий программный
код (а можно придумать свой):

```python
numeric_dtypes = ['int64', 'float64']
numerics = [c for c in df.columns if df[c].dtype in numeric_dtypes]
```

```python
df_num = df[numerics]        # исходный df не трогаем
print('Перед dropna:', df_num.shape)
print(df_num.isna().sum().sort_values(ascending=False).head(10))
```

4. Удалите столбец Id и пропущенные значения.

```python
# Удаляем Id и полностью пустой столбец
del df_num['Id']
del df_num['Unnamed: 81']
```

```python
df_num = df_num.dropna()
print('После dropna:', df_num.shape)
df_num.info()
```

5. Разделите набор данных на входные данные X (все столбцы кроме SalePrice) и ответы y (столбец SalePrice).

```python
X = df_num.drop('SalePrice', axis = 1)
y = df_num['SalePrice']

print('X shape:', X.shape)
print('y shape:', y.shape)
print('Первые 5 строк X:')
print(X.head())
print('Первые 5 значений y:')
print(y.head())
```

6. Разделите данные на обучающую и тестовую выборки.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

7. Обучите модель LassoCV (установите значение гиперпараметра cv
самостоятельно). Оцените качество полученной модели. Посмотрите на
коэффициенты модели. Есть ли коэффициенты равные 0? Что это означает? 

```python
lasso5 = LassoCV(cv=5) # Для примера возьмем cv = 5
```

Это означает, что для подбора alpha будет использоваться 5-блочная перекрестная проверка.

```python
lasso5.fit(X_train, y_train) # Обучаем
lasso5.score(X_test, y_test) # Оцениваем качество
lasso5.coef_ # Значения коэффициентов
```

```python
lasso5.score(X_test, y_test) # Оцениваем качество
```

Значение R2 ≈ 0.75 означает, что модель объясняет примерно 75% вариации целевой переменной SalePrice на тестовых данных, то есть предсказывает достаточно хорошо, но остаётся заметная часть разброса, которую линейная модель с L1‑регуляризацией не улавливает.

LassoCV — это линейная регрессия с L1‑регуляризацией, где параметр штрафа alpha подбирается по кросс‑валидации (cv=5, то есть 5‑блочная перекрёстная проверка).​

L1‑штраф стремится «обнулить» как можно больше коэффициентов, поэтому Lasso даёт разреженное решение: многие веса становятся ровно нулевыми.​

Ненулевые коэффициенты означают, что соответствующие признаки реально используются моделью: изменение признака влияет на прогноз цены (в плюс или минус, в зависимости от знака).​

Коэффициенты, равные 0, означают, что эти признаки модель посчитала несущественными — их влияние либо очень маленькое, либо дублируется другими признаками, и Lasso полностью исключило их из итоговой модели. Это по сути встроенный отбор признаков.
