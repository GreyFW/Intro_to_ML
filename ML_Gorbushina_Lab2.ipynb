## (3 балла)
Напишите программы, которые при помощи matplotlib выводят следующие графики ...

```python
import matplotlib.pyplot as plt
import numpy as np

def graph1():
    plt.figure(figsize=(8, 6))
    x_axis = [x*10 for x in range(0, 6)]
    y_axis = [y*30 for y in range(0, 6)]
    
    plt.xlabel('x - axis')
    plt.ylabel('y - axis')
    plt.title('Draw a line')
    plt.plot(x_axis, y_axis, color='blue')
    plt.xlim(0, 50)
    plt.ylim(0, 160)
    plt.show()

def graph2():
    plt.figure(figsize=(8, 6))
    
    x_blue = np.array([10, 15, 20, 25, 30])
    y_blue = np.array([20, 30, 40, 25, 10])
    
    x_red = np.array([10, 15, 20, 25, 30])
    y_red = np.array([40, 25, 10, 20, 30])
    
    plt.title("Two or more lines with different widths and colors with suitable legends")
    plt.xlabel('x - axis')
    plt.ylabel('y - axis')
    plt.xlim(10, 30)
    plt.ylim(10, 40)
    plt.plot(x_blue, y_blue, color='blue', linewidth=3, label='line1-width-3')
    plt.plot(x_red, y_red, color='red', linewidth=5, label='line2-width-5')
    plt.legend()
    
    plt.show()

    
def graph3():
    plt.figure(figsize=(8, 6))
    
    x_blue = np.array([10, 15, 20, 25, 30])
    y_blue = np.array([20, 30, 40, 25, 10])
    
    x_red = np.array([10, 15, 20, 25, 30])
    y_red = np.array([40, 25, 10, 20, 30])
    
    plt.title("Plot with two or more lines with different styles")
    plt.xlabel('x - axis')
    plt.ylabel('y - axis')
    plt.xlim(10, 30)
    plt.ylim(10, 40)
    plt.plot(x_blue, y_blue, color='blue', linewidth=3, label='line1-dotted', linestyle=':')
    plt.plot(x_red, y_red, color='red', linewidth=5, label='line2-dashed', linestyle='--', dashes=(1, 1))
    plt.legend()
    
    plt.show()

def graph4():
    plt.figure(figsize=(8, 6))
    
    x = [1, 4, 5, 6, 7]
    y = [2, 6, 3, 6, 3]
    
    plt.title("Display marker")
    plt.xlabel('x - axis')
    plt.ylabel('y - axis')
    plt.xlim(1, 8)
    plt.ylim(1, 8)
    plt.plot(x, y, color='red', linewidth=2, linestyle='-.', dashes=(2, 1, 1, 1),
             marker='o', markerfacecolor='blue', markeredgecolor='blue', markersize=10)
    plt.show()
    

def graph5():
    plt.figure(figsize=(8, 6))
    
    x_blue = [2, 3, 5, 6, 8]
    y_blue = [1, 5, 10, 18, 20]
    
    x_red = [3, 4, 6, 7, 9]
    y_red = [3, 6, 11, 20, 22]
    
    plt.xlim(0, 10)
    plt.ylim(0, 30)
    plt.plot(x_red, y_red, linestyle='none', marker='o', color='red', markeredgecolor='black', markersize=6)
    plt.plot(x_blue, y_blue, linestyle='none', marker='*', color='blue', markersize=4)
    plt.show()
    

def graph6():
    dates_x = ['2016-10-03', '2016-10-04', '2016-10-05', '2016-10-06', '2016-10-07']
    y = [772.5, 776.4, 776.45, 776.85, 775.08]
    
    plt.figure(figsize=(8, 6))
    
    plt.title('Closing stock value of Alphabet Inc.')
    plt.xlabel('Date')
    plt.ylabel('Closing Value')
    plt.ylim(772.5, 777.0)
    plt.xlim()
    plt.plot(dates_x, y, 'o-', markersize=6, linewidth=2,
             color='red', markerfacecolor='red', markeredgecolor='black')
    plt.grid(True, which='major', alpha=0.5, linestyle='-', linewidth=0.8, color='red')
    plt.grid(True, which='minor', alpha=0.2, linestyle='-.', linewidth=0.5, color='gray')
    plt.legend()
    plt.minorticks_on()

    plt.show()

graph1()
graph2()
graph3()
graph4()
graph5()
graph6()
```

```python
def task_2():
    plt.figure(figsize=(8, 8))

    x_blue = np.array([10, 15, 20, 25, 30])
    y_blue = np.array([20, 30, 40, 25, 10])
    
    x_red = np.array([10, 15, 20, 25, 30])
    y_red = np.array([40, 25, 10, 20, 30])

    plt.title("Элементы изображения")
    plt.xlabel('Подпись оси OX')
    plt.ylabel('Подпись оси OY')
    
    plt.xlim(10, 30)
    plt.ylim(10, 40)
    
    plt.minorticks_on()
    plt.tick_params(which='both', direction='out')
    
    plt.grid(True, alpha=0.5, linestyle='--', linewidth=1, color='gray')    
    plt.plot(x_blue, y_blue, color='blue', linewidth=3, label='Синяя линия', 
             marker='o', markerfacecolor='blue', markersize=8)
    plt.plot(x_red, y_red, color='red', linewidth=3, label='Красная линия',
             marker='s', markerfacecolor='red', markersize=8)
    
    plt.legend()
    
    plt.show()

task_2()
```
